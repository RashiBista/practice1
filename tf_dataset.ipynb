```python
import tensorflow as tf
```

```python
daily_sales_number=[101,21,-1,20,11,23]
```

```python
tf_dataset= tf.data.Dataset.from_tensor_slices(daily_sales_number)
```

```python
tf_dataset
```

```python
for sales in tf_dataset:
    print(sales.numpy())
```

```python
for sales in tf_dataset.take(3):
  print(sales.numpy())
```

```python
tf_dataset=tf_dataset.filter(lambda x:x>0)
```

```python
tf_dataset
```

```python
for sales in tf_dataset.as_numpy_iterator():
  print(sales)
```

```python
tf_dataset=tf_dataset.map(lambda x:x*154)
for sales in tf_dataset.as_numpy_iterator():
  print(sales)
```

```python
tf_dataset=tf_dataset.shuffle(3)
for sales in tf_dataset.as_numpy_iterator():
  print(sales)
```

```python
tf_dataset=tf_dataset.batch(2)
for sales in tf_dataset.as_numpy_iterator():
  print(sales)
```

```python
tf_dataset= tf.data.Dataset.from_tensor_slices(daily_sales_number)
```

```python
tf_dataset=tf_dataset.filter(lambda x:x>0).map(lambda y:y*154).shuffle(3).batch(2)
for sales in tf_dataset.as_numpy_iterator():
  print(sales)
```
