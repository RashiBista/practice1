```python
import pickle
```

```python
import json
```

```python
class Fruit:
  def __init__(self,name: str,calories: float):
    self.name = name
    self.calories = calories
  def describe_fruit(self):
    print(self.name,self.calories, sep= '')
if __name__=='__main__':
  fruit: Fruit= Fruit('Banana',100)
  fruit.describe_fruit()

  fruit.calories = 150

  with open('banana.jason','w') as file:
    data={'name': fruit.name,'calories': fruit.calories}
    json.dump(data,file) #kind of like writing

```

```python
  with open('banana.jason','r') as file:
    data = json.load(file) #kind of like reading
    print(data)
```

```python
##using pickle
with open('bannana.pickle', 'wb') as file:
  pickle.dump(fruit,file)
```
